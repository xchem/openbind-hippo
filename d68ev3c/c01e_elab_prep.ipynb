## Config

````python
target_name = "D68EV3C"
target_dir = Path(environ["BULK"]) / "TARGETS" / target_name
cycle_name = "cycle_01"
cycle_dir = Path(cycle_name)
aligned_dir = target_dir / "aligned_files"
````

## Animal

````python
animal = hippo.HIPPO(target_name, target_dir / f"{target_name}.sqlite")
````

## Load chem review

````python
df = pd.read_excel("cycle_01/chem_review/openbind_d68ev3c_c1_scaffolds_1steps.xlsx")#, sheet_name="chem-review")
select = df[df["lisa-select"]=="Y"]
print(len(select))
ids = [int(cname[1:]) for cname in select["target-names"]]
cset = animal.compounds[ids]
cset.add_tag("openbind_d68ev3c_c1_scaffolds_chemok")
cset
````

````python
df = pd.read_excel("cycle_01/chem_review/openbind_d68ev3c_c1_scaffolds_2steps.xlsx")#, sheet_name="chem-review")
select = df[df["lisa-select"]=="Y"]
print(len(select))
ids = [int(cname[1:]) for cname in select["target-names"]]
cset += animal.compounds[ids]
cset.add_tag("openbind_d68ev3c_c1_scaffolds_chemok")
cset
````

````python
df = pd.read_excel("cycle_01/chem_review/openbind_d68ev3c_c1_scaffolds_3steps.xlsx")#, sheet_name="chem-review")
select = df[df["lisa-select"]=="Y"]
print(len(select))
ids = [int(cname[1:]) for cname in select["target-names"]]
cset += animal.compounds[ids]
cset.add_tag("openbind_d68ev3c_c1_scaffolds_chemok")
cset
````

## Quote reactants

## In-Stock recipe
