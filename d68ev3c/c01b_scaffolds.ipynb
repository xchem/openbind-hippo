# Config

````python
target_name = "D68EV3C"
target_dir = Path(environ["BULK"]) / "TARGETS" / target_name
cycle_name = "cycle_01"
cycle_dir = Path(cycle_name)
aligned_dir = target_dir / "aligned_files"
````

## Animal

````python
animal = hippo.HIPPO(target_name, target_dir / f"{target_name}.sqlite")
````

# Annotate Pockets (subsites)

Assign subsites from Canonical site definitions in metadata (for all experimental hits)

````python
hits = animal.poses(tag="hits")
hits.set_subsites_from_metadata_field()
````

Assign subsites of derivatives based on inspiration subsites

````python
animal.db.set_derivative_subsites()
````

Print a summary

````python
hits.derivatives.subsite_summary();
````

# Tag Scaffolds

Tag your scaffolds, e.g. by their BulkDock export:

````python
scaffold_poses = (
    animal.poses.get_by_metadata_substring_match("openbind_d68ev3c_c1_fragmenstein_1669J2U.sdf") + 
    animal.poses.get_by_metadata_substring_match("openbind_d68ev3c_c1_knitwork_pure_1669ISE.sdf") + 
    animal.poses.get_by_metadata_substring_match("openbind_d68ev3c_c1_knitwork_impure_1669BTJ.sdf")
)

scaffold_poses.add_tag("openbind_D68EV3C_01_scaffolds")
scaffold_poses
````
