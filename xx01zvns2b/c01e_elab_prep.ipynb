## Scaffold recipe scoring

- Use the `calculate_interactions.py` SLURM script to calculate interactions

````python
# restrict the poses being evaluated on
pose_pool = animal.poses.get_by_metadata_substring_match("GNINA pK")
pose_pool
````

````python
scorer = hippo.Scorer(
    animal.db, 
    "cycle_01/rgen/xx01zvns2b_c1_scaffolds_recipes", 
    attributes=[
        "num_compounds", 
        "num_reaction_types",
    ], 
    load_cache=False, 
    populate=True,
    allowed_poses=pose_pool,
)

### Custom GNINA pK scoring

def avg_GNINA_pK(recipe):
    poses = recipe.poses
    metadata_lookup = poses.pose_id_metadata_dict
    values = [d.get("GNINA pK") for d in metadata_lookup.values()]
    values = [pK for pK in values if pK is not None]
    return sum(values) / len(values)

scorer.add_custom_attribute("avg_GNINA_pK", avg_GNINA_pK)

scorer.weights = [1,3,1]

scorer
````

````python
scorer.summary()
````

````python
scorer.plot(["num_compounds", "avg_GNINA_pK"])
````

````python
# Recipe_1GEXTRS
proposal = scorer.best
proposal
````

````python
proposal.write_CAR_csv("cycle_01/rgen/xx01zvns2b_c1_scaffolds_recipes/Recipe_1GEXTRS_CAR.csv")
proposal.write_reactant_csv("cycle_01/rgen/xx01zvns2b_c1_scaffolds_recipes/Recipe_1GEXTRS_reactants.csv")
````
