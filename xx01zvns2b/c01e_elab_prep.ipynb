## Generate EUR 10k scaffold recipes

````python
route_pool = hippo.RouteSet.from_product_ids(animal.db, mod_recipe.product_compounds.ids)
route_pool
````

````python
data = []
for route in mrich.track(route_pool.prune_unavailable(["Enamine"])):
    data.append(dict(product_id=route.product.id, route_id=route.id, chemistry=route.reactions.types[0], price=route.price.amount))
df = pd.DataFrame(data)
df
````

````python
df_min = df.loc[df.groupby('chemistry')['price'].idxmin()]
df_min = df_min.sort_values("price", ascending=False)
print(df_min["price"].sum())
df_min
````

````python
# Starting recipe representing all chemistries
starting_recipe = hippo.Recipe.from_compounds(animal.compounds[list(df_min["product_id"].unique())])
starting_recipe
````

````python
starting_recipe.write_json("cycle_01/rgen/openbind_xx01zvns2b_c1_scaffold_rgen_starting_recipe.json")
starting_recipe.write_CAR_csv("cycle_01/rgen/openbind_xx01zvns2b_c1_scaffold_rgen_starting_recipe_CAR.csv")
````

````python
gen = hippo.RandomRecipeGenerator(
    db=animal.db, 
    start_with=starting_recipe, 
    suppliers=['Enamine'],
    route_pool=route_pool,
    out_key="cycle_01/rgen/xx01zvns2b_c1_scaffolds"
)
````

````python
# test the genenerator
recipe = gen.generate(
    budget = 10_000,
    currency = "EUR",
)
````
