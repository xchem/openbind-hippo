# Fix incorrect references

```python
sql = """
SELECT pose_id, pose_alias, pose_reference FROM pose
WHERE pose_alias LIKE '%C%-P%'
"""

records = animal.db.execute(sql).fetchall()
len(records)
```

```python
update = []
for record in mrich.track(records):
    pose_id, alias, ref_id = record
    correct_ref = int(alias.split("-P")[-1])
    if ref_id != correct_ref:
        # print(pose_id, alias, ref_id, correct_ref)
        update.append((pose_id, correct_ref))
len(update)
```

```python
sql = """
UPDATE pose
SET pose_reference = ?2
WHERE pose_id = ?1
"""

if update:
    animal.db.executemany(sql, update)
    animal.db.commit()
```

```python
sql = """
SELECT pose_id, pose_reference, pose_path FROM pose
WHERE pose_path LIKE '%.fake.mol'
"""

c = animal.db.execute(sql)
records = c.fetchall()
len(records)
```

```python
update = []
for pose_id, ref_id, pose_path in records:
    correct_ref = int(pose_path.split("-P")[-1].removesuffix(".fake.mol"))
    if correct_ref != ref_id:
        # print(pose_id, ref_id, correct_ref)
        update.append((pose_id, correct_ref))
len(update)
```

```python
sql = """
UPDATE pose
SET pose_reference = ?2
WHERE pose_id = ?1
"""

if update:
    animal.db.executemany(sql, update)
    animal.db.commit()
```
