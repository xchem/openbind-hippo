# Fix incorrect pose reference IDs

````python
sql = """
SELECT pose_id, pose_reference, pose_path FROM pose
WHERE pose_path LIKE '%.fake.mol'
"""

c = animal.db.execute(sql)
records = c.fetchall()
len(records)
````

````python
update = []
for pose_id, ref_id, pose_path in records:
    correct_ref = int(pose_path.split("-P")[-1].removesuffix(".fake.mol"))
    if correct_ref != ref_id:
        # print(pose_id, ref_id, correct_ref)
        update.append((pose_id, correct_ref))
len(update)
````

````python
sql = """
UPDATE pose
SET pose_reference = ?2
WHERE pose_id = ?1
"""

if update:
    animal.db.executemany(sql, update)
    animal.db.commit()
````

# Export for co-folding

````python
pset = animal.poses(tag="openbind_xx01zvns2b_c1_scaffolds")
pset
````

````python
sql = f"""
SELECT pose_id, pose_smiles, pose_reference FROM pose
WHERE pose_id IN {pset.str_ids}
"""

cursor = animal.db.execute(sql)

data = []
for pose_id, pose_smiles, pose_reference in cursor:
    data.append(dict(id=pose_id, smiles=pose_smiles, reference_id=pose_reference))
df = pd.DataFrame(data)

df = df.set_index("id")

df.head()
````

````python
pattern = re.compile(r"^.*-x[0-9]{4}_([A-Z])_[0-9]*_v[0-9]*$")

def get_sequence(pose_id):
    p = animal.poses[pose_id]
    sys = p.protein_system
    if not sys:
        mrich.error("No protein system")
        return None

    match = pattern.match(p.metadata["observation_longname"])
    ligand_chain = match.group(1)

    try:
        chain = sys.get_chain(ligand_chain)
        return chain.sequence
    except:
        seqs = set()
        for chain in sys.chains:
            seq = chain.sequence
            seqs.add(seq)
        
        if len(seqs) > 1:
            mrich.warning("Multiple sequences!", p)
        
        return list(seqs)[0]
````

````python
lookup = {}
for reference_id in mrich.track(set(df["reference_id"].values)):
    i = int(reference_id)
    lookup[i] = get_sequence(i)
````

````python
lookup[86] = animal.P86.protein_system.get_chain("B").sequence
````

````python
df["sequence"] = df["reference_id"].apply(lambda x: lookup[x])
````

````python
df = df.drop(columns=["reference_id"])
df.to_csv("openbind_xx01zvn_c1_scaffold_smiles_sequences.csv")
````
