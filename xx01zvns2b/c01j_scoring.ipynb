```python
scaffold_poses = animal.poses(tag="openbind_xx01zvns2b_c1_scaffolds_chemok_bbok")
elaborated_scaffolds = scaffold_poses.compounds.elabs.scaffolds
elaborated_scaffold_poses = animal.poses[set(elaborated_scaffolds.poses.ids).intersection(set(scaffold_poses.ids))]
elaborated_scaffold_poses
```

```python
data = []
for i,pose in enumerate(elaborated_scaffold_poses):
    print(i, pose)
    compound = pose.compound
    elabs = compound.elabs
    reference = pose.reference
    inspirations = pose.inspirations

    inspiration_d = dict()
    for i, name in enumerate(inspirations.names):
        inspiration_d[f"hit{i+1}"] = name
    
    for elab in mrich.track(elabs):
        d = dict(smiles=elab.smiles)
        d.update(inspiration_d)
        data.append(d)

df = pd.DataFrame(data)
df.to_csv("cycle_01/syndirella/elabs/xx01zvns2b_c1_elab_bulkdock_input.csv", index=False)
df
```

```python
from pandas import DataFrame

sql = f"""
SELECT tag_name,
COUNT(DISTINCT tag_compound),
COUNT(DISTINCT tag_pose)
FROM tag
WHERE tag_compound IN {animal.scaffolds.str_ids}
GROUP BY tag_name
ORDER BY tag_name
"""

cursor = animal.db.execute(sql)

data = [
    dict(tag=a, num_compounds=b, num_poses=c) for a, b, c in cursor.fetchall()
]

df = DataFrame(data)
df = df.set_index("tag")

# compounds with poses

sql = f"""
SELECT tag_name, COUNT(DISTINCT pose_compound) FROM tag
INNER JOIN pose
ON tag_pose = pose_id
WHERE pose_compound IN {animal.scaffolds.str_ids}
GROUP BY tag_name
ORDER BY tag_name
"""

cursor = animal.db.execute(sql)

for tag, count in cursor.fetchall():
    df.loc[tag, "num_posed_compounds"] = count

df = df.fillna(0)
df = df.astype(int)

df
```

## Get GNINA scores

```python
ref_sdf = "cycle_01/syndirella/elabs/openbind_xx01zvns2b_c1_elab_chemspace_filtered.sdf"
pdf = PandasTools.LoadSDF(ref_sdf)
pdf = pdf[["HIPPO Pose ID", "ID"]].rename(columns={"HIPPO Pose ID":"id", "ID":"name"}).iloc[1:].set_index("name")
pdf.head()
```

```python
sdf = pd.read_csv("cycle_01/gnina/openbind_xx01zvns2b_c1_elab_chemspace_filtered_scores.csv")
sdf.head()
```

```python
sdf["pose_id"] = sdf["name"].apply(lambda x: int(pdf.loc[x, "id"]))
sdf = sdf.set_index("pose_id")
sdf
```

```python
pose_ids = set()
for i,row in sdf.iterrows():
    pose = animal.poses[i]
    metadata = pose.metadata
    metadata.update({
        "gnina_score":row["gnina_score"],
        "gnina_score_min":row["gnina_score_min"],
        "smina_score_min":row["smina_score_min"],
        "rmsd(smina/fragmenstein)":row["min_rmsd"],
    })
    pose_ids.add(i)
    # break
scored_poses = animal.poses[pose_ids]
scored_poses.add_tag("openbind_xx01zvns2b_c1_elab_chemspace_filtered_scored")
```

## GNINA score distribution

```python
scored_poses = animal.poses(tag="openbind_xx01zvns2b_c1_elab_chemspace_filtered_scored")
scored_poses
```

```python
data = []
for pose in scored_poses:
    metadata = pose.metadata
    data.append({
        "pose_id":pose.id, 
        "gnina_score":metadata["gnina_score"],
        "gnina_score_min":metadata["gnina_score_min"],
        "smina_score_min":metadata["smina_score_min"],
        "rmsd(smina/fragmenstein)":metadata["rmsd(smina/fragmenstein)"],
    })
df = pd.DataFrame(data)
```

```python
score_cutoff = df["gnina_score_min"].quantile(0.60)
score_cutoff
```

```python
px.histogram(df, x="gnina_score_min")
```

```python
filtered_scored_ids = df[df["gnina_score_min"] >= score_cutoff]["pose_id"].unique()
```

```python
filtered_scored_poses = animal.poses[filtered_scored_ids]
filtered_scored_poses
```
