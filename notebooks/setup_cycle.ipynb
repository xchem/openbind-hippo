```python
# imports
from fragalysis.widgets import download
from pathlib import Path
import shutil
from os import environ
import json
```

```python
# this is the destination you should download to:
bulk_targets_dir = Path(environ["BULK"]) / "TARGETS"
print(bulk_targets_dir)
```

```python
download()
```

## Initialise directories for new cycle

```python
target_name = "XX01ZVNS2B" # as in Fragalysis
cycle_number = 1
```

```python
# create directories
target_dir = Path("..") / target_name.lower()
cycle_dir = target_dir / f"cycle_{cycle_number:02}"
cycle_dir.mkdir(parents=True, exist_ok=True)
```

```python
# copy aligned files
shutil.copytree(bulk_targets_dir / target_name / "aligned_files", target_dir / "aligned_files")
```

## Copy template notebook(s)

```python
def copy_notebook_with_replacement(src_path, dst_path):
    # Read the original notebook
    with open(src_path, 'r', encoding='utf-8') as f:
        notebook = json.load(f)
    
    # Replace text in each cell
    for cell in notebook.get('cells', []):
        if 'source' in cell:
            new_source = []
            for line in cell['source']:
                new_line = line.replace("TARGET_NAME", target_name)
                new_line = line.replace("TNAME", target_name.lower())
                new_line = new_line.replace("CYCLE_NUMBER", f"{cycle_number:02}")
                new_line = new_line.replace("CNUM", f"{cycle_number}")
                new_source.append(new_line)
            cell['source'] = new_source

    # Write the modified notebook to the new path
    with open(dst_path, 'w', encoding='utf-8') as f:
        json.dump(notebook, f, indent=1, ensure_ascii=False)
```

```python
copy_notebook_with_replacement("template_a_merging.ipynb", target_dir / f"c{cycle_number:02}a_merging.ipynb")
```

```python
copy_notebook_with_replacement("template_b_scaffolds.ipynb", target_dir / f"c{cycle_number:02}b_scaffolds.ipynb")
```
