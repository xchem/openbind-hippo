# Load Retrosynthesis Outputs

````python
animal.add_syndirella_routes(
    "cycle_CYCLE_NUMBER/syndirella/openbind_TNAME_cCNUM_scaffolds_syndirella_input.pkl.gz", 
    CAR_only=False, 
    check_chemistry=False,
)
````

````python
animal.db.prune_duplicate_routes()
````

````python
cset = animal.poses(tag="openbind_TNAME_cCNUM_scaffolds").compounds
cset.add_tag("openbind_TNAME_cCNUM_scaffolds")
````

````python
recipe = hippo.Recipe.from_compounds(
    cset, 
    use_routes=True, 
    pick_cheapest=False, 
    pick_first=True
)
````

````python
recipe.write_json("cycle_CYCLE_NUMBER/syndirella/retrosynthesis/openbind_TNAME_cCNUM_scaffolds.json")
````
