```python
paths = {
    "xx01zvns2b":"../xx01zvns2b/cycle_02/analogues/openbind_xx01zvns2b_c2_hit_analogues_85percent.csv",
    "a71ev2a":"../a71ev2a/cycle_02/analogues/openbind_a71ev2a_c2_hit_analogues_85percent.csv",
    "d68ev3c":"../d68ev3c/cycle_02/analogues/openbind_d68ev3c_c2_hit_analogues_85percent.csv",
}
```

```python
df = pd.read_csv(paths["a71ev2a"]).fillna(False)
df.head()
```

```python
# series = {key:pd.read_csv(path).set_index("smiles")[[
#     f"openbind_{key}_c2_hit_analogues_85percent",
#     f"openbind_{key}_c2_hit_analogues_90percent",
#     f"openbind_{key}_c2_hit_analogues_95percent",
# ]] for key,path in paths.items()}

series = {}

for key,path in paths.items():

    df = pd.read_csv(path).set_index("smiles").fillna(False)

    if "soaks" in df.columns:
        df = df[df["soaks"] == False]

    series[key] = df[[
        f"openbind_{key}_c2_hit_analogues_85percent",
        f"openbind_{key}_c2_hit_analogues_90percent",
        f"openbind_{key}_c2_hit_analogues_95percent",
    ]]
```

```python
for k in series:
    if not series[k].index.is_unique:
        series[k] = series[k][~series[k].index.duplicated()]
```

```python
series
```

```python
df = pd.concat(series, axis=1)
```

```python
df = df.fillna(False)
```
