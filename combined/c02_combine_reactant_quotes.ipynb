## Combine Cycle 02 elab reactants into a single quote

```python
import pandas as pd
import mrich
```

```python
paths = {
    "xx01zvns2b":"../xx01zvns2b/cycle_02/syndirella/elabs/openbind_xx01zvns2b_c2_elab_route_reactants.csv",
    "a71ev2a":"../a71ev2a/cycle_02/syndirella/elabs/openbind_a71ev2a_c2_elab_route_reactants.csv",
    "d68ev3c":"../d68ev3c/cycle_02/syndirella/elabs/openbind_d68ev3c_c2_elab_route_reactants.csv",
}
```

```python
series = {key:pd.read_csv(path).set_index(["smiles"])["id"].astype(int) for key,path in paths.items()}
```

```python
for k in series:
    if not series[k].index.is_unique:
        series[k] = series[k][~series[k].index.duplicated()]
```

```python
series
```

```python
df = pd.concat(series, axis=1)
```

```python
df.columns = [f"HIPPO Compound ID ({c})" for c in df.columns]
```

```python
for col in df.columns:
    df[col] = df[col].astype("Int64")
```

```python
mrich.print(df)
```

```python
df.to_csv("openbind_c2_combined_elab_reactants.csv")
```
