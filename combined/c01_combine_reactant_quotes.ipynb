## Combine Cycle 01 chemspace reactants into a single quote

```python
import pandas as pd
import mrich
```

```python
paths = {
    "xx01zvns2b":"../xx01zvns2b/cycle_01/syndirella/elabs/openbind_xx01zvns2b_c1_all_elabs_reactants.csv",
    "a71ev2a":"../a71ev2a/cycle_01/rgen/openbind_a71ev2a_c1_elabs_filtered_reactants.csv",
    "d68ev3c":"../d68ev3c/cycle_01/syndirella/elabs/openbind_d68ev3c_c1_elab_chemspace_reactants.csv",
}
```

```python
series = {key:pd.read_csv(path).set_index(["quote_entry", "quoted_smiles"])["compound_id"].astype(int) for key,path in paths.items()}
```

```python
for k in series:
    if not series[k].index.is_unique:
        series[k] = series[k][~series[k].index.duplicated()]
```

```python
series
```

```python
df = pd.concat(series, axis=1)
```

```python
df.columns = [f"HIPPO Compound ID ({c})" for c in df.columns]
```

```python
for col in df.columns:
    df[col] = df[col].astype("Int64")
```

```python
mrich.print(df)
```

```python
df.to_csv("openbind_c1_combined_chemspace_reactants.csv")
```
